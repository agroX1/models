```python
def clean_product_column(product_str):
    if isinstance(product_str, str):
        product_list = []
        for product in product_str.split(','):
            cleaned_product = product.split('×')[-1].strip()
            cleaned_product = cleaned_product.split('|')[0].strip()
            product_list.append(cleaned_product)
        return product_list
    else:
        return product_str

transaction_data['Cleaned_Products'] = transaction_data['Product(s)'].apply(clean_product_column)

max_products = transaction_data['Cleaned_Products'].apply(lambda x: len(x) if isinstance(x, list) else 0).max()

for i in range(1, max_products + 1):
    transaction_data[f'Product_{i}'] = transaction_data['Cleaned_Products'].apply(lambda x: x[i-1] if isinstance(x, list) and len(x) >= i else None)

transaction_data.head()
```

## Data preparation


Prepare the `transaction_data` for building a recommendation model. This may involve tasks such as creating a user-item matrix or other suitable data structures.


**Reasoning**:
Prepare the data for the recommendation model by creating a user-item matrix. This involves selecting relevant columns, melting the data, handling missing values, and creating a pivot table.



```python
product_data = transaction_data_clean[['Customer_ID'] + [f'Product_{i}' for i in range(1, 23)]]

melted_product_data = product_data.melt(id_vars=['Customer_ID'], value_name='Product')

melted_product_data.dropna(subset=['Product'], inplace=True)

user_item_matrix = melted_product_data.pivot_table(index='Customer_ID', columns='Product', aggfunc='size', fill_value=0)

display(user_item_matrix.head())
```

## Model selection

Choose an appropriate recommendation model based on the data and the goal of generating personalized recommendations.


**Reasoning**:
Based on the binary nature of the user-item matrix and the goal of generating personalized recommendations using implicit feedback, a suitable collaborative filtering model is needed. Matrix Factorization models like Singular Value Decomposition (SVD) are well-suited for this as they can capture latent features that explain the observed purchases and predict potential future purchases. Since we are dealing with implicit feedback (purchase or no purchase), techniques like Funk SVD or algorithms designed for implicit data such as LightFM could be considered. Given the available data in the form of a user-item matrix, implementing a basic SVD-based approach will be appropriate to derive latent factors for users and items.



```python
print("A suitable recommendation model for this binary user-item matrix with implicit feedback is a Matrix Factorization model, such as Singular Value Decomposition (SVD).")
print("SVD works by decomposing the user-item matrix into two lower-rank matrices, one representing latent features for users and the other for items.")
print("These latent features capture underlying preferences of users and characteristics of items that explain the observed purchase behavior.")
print("For implicit feedback data like this, where the values represent purchases (1) and non-purchases (0), SVD-based methods can learn to predict the likelihood of a user interacting with an item they haven't purchased yet.")
print("This aligns well with the goal of generating personalized product recommendations by identifying items a user is likely to purchase based on their past behavior and the behavior of similar users.")
```

```python
product_data = transaction_data_clean[['Customer_ID'] + [f'Product_{i}' for i in range(1, 23)]]

melted_product_data = product_data.melt(id_vars=['Customer_ID'], value_name='Product')

melted_product_data.dropna(subset=['Product'], inplace=True)

melted_product_data['Purchase'] = 1

from surprise import Dataset, Reader, SVD

reader = Reader(rating_scale=(0, 1))
data = Dataset.load_from_df(melted_product_data[['Customer_ID', 'Product', 'Purchase']], reader)

trainset = data.build_full_trainset()

model = SVD()

model.fit(trainset)
```

## Prediction

Use the trained model to generate personalized product recommendations for each customer.


**Reasoning**:
Get unique customer IDs and product names, and then iterate through each customer and each product they haven't purchased to predict the purchase likelihood using the trained SVD model. Store these predictions in a list of tuples and convert to a pandas DataFrame.



```python
unique_customer_ids = user_item_matrix.index.unique()
unique_product_names = user_item_matrix.columns.unique()

predictions = []
for customer_id in unique_customer_ids:
    purchased_products = user_item_matrix.loc[customer_id][user_item_matrix.loc[customer_id] > 0].index.tolist()
    for product_name in unique_product_names:
        if product_name not in purchased_products:
            predicted_likelihood = model.predict(customer_id, product_name, r_ui=0).est
            predictions.append((customer_id, product_name, predicted_likelihood))

predictions_df = pd.DataFrame(predictions, columns=['Customer_ID', 'Product', 'Predicted_Likelihood'])
display(predictions_df.head())
```

**Reasoning**:
For each customer, sort the predicted products by their likelihood in descending order and select the top N recommendations (N=10), then combine the top recommendations for all customers into a single DataFrame.



```python
top_n_recommendations = predictions_df.groupby('Customer_ID').apply(lambda x: x.nlargest(10, 'Predicted_Likelihood')).reset_index(drop=True)
display(top_n_recommendations.head())
```
